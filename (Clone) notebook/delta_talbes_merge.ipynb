```sql
MERGE INTO dev.bronze.emp e
USING dev.bronze.emp_updates u
ON e.emp_id = u.emp_id
WHEN MATCHED THEN
UPDATE SET e.salary = u.salary,
e.is_active = 'Y'
WHEN NOT MATCHED THEN INSERT (
    e.emp_id,
    e.emp_name,
    e.dept_code,
    e.salary,
    e.is_active
)
values(
    u.emp_id,
    u.emp_name,
    u.dept_code,
    u.salary,
    'Y'
)
WHEN NOT MATCHED BY SOURCE THEN 
UPDATE SET e.is_active = 'N'
;
```
