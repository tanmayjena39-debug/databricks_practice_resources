```sql
--ctas for delta table
create table dev.bronze.emp_ctas as select * from dev.bronze.emp ;
```

```sql
select * from dev.bronze.emp_ctas
```

```sql
--deep clone in delta table create entire copy of the table including metadata , partitions and so on.
--but ctas will do the same but the metadata or partitions can be loose

create table dev.bronze.emp_deep_clone deep clone dev.bronze.emp;
```

```sql
DESCRIBE HISTORY dev.bronze.emp_deep_clone
```

```sql
create table dev.bronze.emp_shallow_clone shallow clone dev.bronze.emp;
```
